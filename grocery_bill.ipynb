```python
n = int(input("enter number of items: "))
total = 0

print("\n***********BILL*********")
print("Item name \t Quantity \t Item price")

for i in range(n):
    name = input("\nenter item name: ")
    quantity = float(input("enter quantity: "))
    price = float(input("enter price per unit: "))
    
    amount = quantity * price
    total = total + amount
    
    # Indented this line so it prints immediately after you type each item
    print(name, "\t \t", quantity, "\t \t", amount)

print("***************************")
print("total amount to be paid:", total)
print("*****************************")
```
